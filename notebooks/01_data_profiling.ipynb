# Data Profiling
## Indonesia Palm Oil Supply Chain

### Tujuan

Notebook ini digunakan untuk melakukan profiling awal terhadap
dataset Indonesia palm oil - Supply chain data dari Trase.

Profiling dilakukan untuk memahami ukuran, struktur, kualitas,
distribusi, serta karakteristik temporal dan geografis dataset
sebelum proses data cleaning.

Hasil profiling digunakan sebagai dasar untuk menentukan proses
cleaning pada notebook `02_data_cleaning.ipynb`.

```python
import polars as pl
import duckdb
```

```python
DATA_PATH = "../data/raw/indonesia_palm_oil_v1_2_5.csv"

DATA_PATH
```

```python
df = pl.scan_csv(DATA_PATH)

df
```

```python
df.collect_schema()
```

```python
row_count = df.select(pl.len()).collect()

row_count
```

```python
column_count = len(df.collect_schema())

column_count
```

```python
df.head(5).collect()
```

```python
null_counts = df.select(
    pl.all().null_count()
).collect()

null_counts
```

```python
year_summary = (
    df
    .group_by("year")
    .agg(pl.len().alias("total_records"))
    .sort("year")
    .collect()
)

year_summary
```
